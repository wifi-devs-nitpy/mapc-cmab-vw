```python
from environment import Environment 
import jax 
import jax.numpy as jnp 
import numpy as np 
from VwContextualBandit import  VwCBAgent
import pandas as pd 
```

```python
env = Environment(n_actions=4, context_size=2, seed=42)
agent = VwCBAgent(
    n_actions = 4, 
    context_size=2
)
```

```python
previous_reward = 0
actions = []
for i in range(10_000):
    context = env.sample_context()
    action = agent.sample(context, previous_reward)
    previous_reward = env.reward(context, action)
    optimal_action = env.optimal_action(context)
    actions.append({
        "context": str(context),
        "action": action, 
        "reward": previous_reward, 
        "optimal_action": optimal_action
    })
```

```python
df = pd.DataFrame(actions)
```
